# NanoAsk GPU benchmark (Colab / Kaggle T4)

1. **Runtime → Change runtime type → T4 GPU** (Colab), or **Accelerator → GPU T4** (Kaggle).
2. Run the cells one at a time (about 10–15 minutes in total). Check that `memory.used` in the first cell is near 0 MiB; if not, restart the session so leftover processes free the GPU. The second cell stops everything if the runtime has no GPU: a CPU runtime fails every GPU test with "Torch not compiled with CUDA enabled".
3. Copy the table printed by the last cell into the README under *GPU results*, and delete the line saying GPU paths are untested.

The benchmark uses a randomly initialised 12-layer/768-dim NanoChat model (throughput does not depend on the weight values), 128 requests with 100–1000-token prompts and 128–256-token outputs, and `max_model_len=2048`.

```
import torch
if not torch.cuda.is_available():
    raise SystemExit("No GPU in this runtime (torch " + torch.__version__ + ").\n"
                     "Colab: Runtime -> Change runtime type -> T4 GPU -> Save, then "
                     "Runtime -> Disconnect and delete runtime, and Run all again.\n"
                     "Kaggle: Settings -> Accelerator -> GPU T4 x2 (needs a phone-verified account).")
print("GPU:", torch.cuda.get_device_name(0), "| torch", torch.__version__)
```
